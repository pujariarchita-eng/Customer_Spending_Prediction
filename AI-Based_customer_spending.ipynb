```python

import pandas as pd

url = "https://raw.githubusercontent.com/MahimaSriK/Mall-Customer-Segmentation-Dataset/main/Mall_Customers.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
print(df.head())
```

```python

# Data cleaning
df.columns = df.columns.str.strip()
df = df.drop_duplicates()
df = df.dropna().copy()

numeric_columns = [
    "Age",
    "Annual Income (k$)",
    "Spending Score (1-100)"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df = df.dropna(subset=numeric_columns)

df = df[
    (df["Age"] > 0) &
    (df["Annual Income (k$)"] > 0) &
    (df["Spending Score (1-100)"].between(1, 100))
].copy()

# Create spending categories
def assign_category(score):
    if score <= 33:
        return "Low"
    elif score <= 66:
        return "Medium"
    else:
        return "High"

df["Spending Category"] = (
    df["Spending Score (1-100)"].apply(assign_category)
)

print("Data cleaning completed.")
print(df.head())
print("\nSpending category counts:")
print(df["Spending Category"].value_counts())
print("\nMissing values:")
print(df.isnull().sum())

```
